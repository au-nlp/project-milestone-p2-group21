```python
import pandas
import numpy as np
from scipy.special import stdtrit
bleurt_scores = pandas.read_csv("combined_smolsent_bleurt_scores_madlad.csv")
```

Since we are measuring the performance of our model as the average BLEURT score over our sample (the test set), we know from the central limit theorem that the average will have a normal distribution. This allows us to calculate a confidence interval using our sample standard deviation. 

We calculate the 95% confidence interval of the average BLEURT score of our models. 

```python
def confidence_interval(name):
    n = bleurt_scores[name].__len__()
    mean = np.mean(bleurt_scores[name])
    std = np.std(bleurt_scores[name], ddof=1)
    J_sigma = stdtrit(n, 0.975) * std / np.sqrt(n)
    print(f"95% confidence interval {name:20s} [{mean - J_sigma:.4f}, {mean + J_sigma:.4f}] - mean {mean:.3f} - std {std:.3f} - {mean:.3f}+-{J_sigma:.3f}")
```

```python
confidence_interval("Baseline_BLEURT")
confidence_interval("Token Swap_BLEURT")
confidence_interval("Fine-tuned_BLEURT")
confidence_interval("TS Fine-tuned_BLEURT")
```
